# 2. Functions and Good Practices

Functions are one of the most important tools for writing maintainable Python programs.

## Learning objectives

By the end of this notebook, you should be able to:

- define functions;
- use parameters and return values;
- distinguish local and global variables;
- use default and keyword arguments;
- write docstrings;
- avoid unnecessary repetition;
- structure code into small reusable units;
- apply basic Python style principles.

## 1. Why functions?

Without functions, code tends to become repetitive.

A function gives a name to a reusable computation.

````python
def greet():
    print("Hello!")

greet()
````

## 2. Parameters

````python
def greet(name):
    print(f"Hello, {name}!")

greet("Anna")
greet("Marco")
````

## 3. Returning values

````python
def square(x):
    return x ** 2

result = square(5)
print(result)
````

`return` sends a value back to the caller.

This is different from `print()`.

````python
def square_print(x):
    print(x ** 2)

def square_return(x):
    return x ** 2

a = square_print(4)
b = square_return(4)

print("a:", a)
print("b:", b)
````

## 4. Multiple parameters

````python
def bmi(weight_kg, height_m):
    return weight_kg / height_m**2

print(bmi(70, 1.75))
````

## 5. Default arguments

````python
def convert_temperature(celsius, decimals=1):
    fahrenheit = celsius * 9 / 5 + 32
    return round(fahrenheit, decimals)

print(convert_temperature(37))
print(convert_temperature(37, decimals=3))
````

## 6. Keyword arguments

Keyword arguments improve readability when a function has several parameters.

````python
print(
    convert_temperature(
        celsius=36.8,
        decimals=2
    )
)
````

## 7. Local scope

````python
def compute_total():
    total = 10 + 20
    return total

print(compute_total())

# The variable `total` only exists inside the function.
````

In general, prefer passing values explicitly to functions rather than relying on many global variables.

## 8. Docstrings

A docstring explains what a function does.

````python
def mean(values):
    """Return the arithmetic mean of a non-empty sequence."""
    return sum(values) / len(values)

print(mean([10, 20, 30]))
print(mean.__doc__)
````

A slightly more informative docstring can describe parameters and return values.

````python
def classify_temperature(temperature, threshold=38.0):
    """
    Classify a temperature measurement.

    Parameters
    ----------
    temperature : float
        Temperature in degrees Celsius.
    threshold : float, default=38.0
        Fever threshold.

    Returns
    -------
    str
        "fever" or "no fever".
    """
    if temperature >= threshold:
        return "fever"
    return "no fever"
````

## 9. Type hints

Type hints make function interfaces easier to understand.

````python
def add(a: float, b: float) -> float:
    return a + b

print(add(2.5, 3.0))
````

Python does not enforce type hints at runtime by default, but editors and static-analysis tools can use them.

## 10. Good practice: one clear responsibility

A function should ideally do one well-defined task.

Less clear:

````python
def process_patient(name, age, temperature):
    print(name)
    print(age)
    if temperature >= 38:
        print("fever")
    else:
        print("no fever")
    return age + 1
````

Better: separate display, classification, and computation.

````python
def has_fever(temperature, threshold=38.0):
    return temperature >= threshold

def next_age(age):
    return age + 1

def patient_summary(name, age, temperature):
    status = "fever" if has_fever(temperature) else "no fever"
    return f"{name}, age {age}, status: {status}"

print(patient_summary("Anna", 40, 38.2))
````

## 11. DRY: Don't Repeat Yourself

````python
def percentage(part, total):
    return 100 * part / total

print(percentage(25, 100))
print(percentage(18, 60))
````

## 12. Guard clauses

Early returns can make logic easier to read.

````python
def safe_divide(a, b):
    if b == 0:
        return None

    return a / b

print(safe_divide(10, 2))
print(safe_divide(10, 0))
````

## 13. Naming

Prefer:

```python
mean_temperature
number_of_patients
calculate_risk_score
```

over:

```python
x
n2
calc
```

unless the short name is obvious from context.

## 14. Keep functions short

A function that becomes very long may be doing too many things.

Splitting complex logic into smaller functions can improve:

- readability;
- testing;
- reuse;
- maintenance.

## 15. Avoid hidden side effects

A function is easier to reason about when its output mainly depends on its inputs.

````python
# Clear and predictable
def normalize(value, minimum, maximum):
    return (value - minimum) / (maximum - minimum)

print(normalize(7, 0, 10))
````

# Possible solution

````python
def summarize_measurements(values: list[float]) -> dict:
    """Return simple summary statistics for a non-empty list of values."""

    if len(values) == 0:
        return {}

    return {
        "min": min(values),
        "max": max(values),
        "mean": sum(values) / len(values),
        "count": len(values),
    }

print(summarize_measurements([10, 12, 15, 9]))
````

# Take-home messages

- Functions make code reusable and easier to organize.
- Prefer returning values over printing inside reusable functions.
- Use clear names, docstrings, and type hints.
- Keep functions focused on one responsibility.
- Avoid unnecessary global state.
- Reduce repeated logic by extracting reusable functions.
