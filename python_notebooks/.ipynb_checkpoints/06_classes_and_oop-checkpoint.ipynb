# 3. Classes and Object-Oriented Programming in Python

Object-oriented programming (OOP) organizes code around **objects** that combine:

- data;
- behavior.

## Learning objectives

By the end of this notebook, you should be able to:

- define a class;
- create objects;
- use attributes and methods;
- understand `self`;
- use constructors;
- define simple properties;
- use inheritance at a basic level;
- understand when OOP is useful.

## 1. From data to objects

A dictionary can represent an entity:

````python
patient = {
    "name": "Anna",
    "age": 42,
    "temperature": 37.1
}

print(patient["name"])
````

A class allows us to combine this data with related behavior.

````python
class Patient:
    pass

p = Patient()
print(p)
````

## 2. The constructor: `__init__`

````python
class Patient:
    def __init__(self, name, age, temperature):
        self.name = name
        self.age = age
        self.temperature = temperature

patient = Patient("Anna", 42, 37.1)

print(patient.name)
print(patient.age)
print(patient.temperature)
````

## 3. What is `self`?

`self` refers to the current object.

When we write:

```python
patient.name
```

Python accesses the `name` attribute stored in that particular object.

## 4. Methods

Methods are functions defined inside a class.

````python
class Patient:
    def __init__(self, name, age, temperature):
        self.name = name
        self.age = age
        self.temperature = temperature

    def has_fever(self, threshold=38.0):
        return self.temperature >= threshold

patient = Patient("Marco", 67, 38.4)

print(patient.has_fever())
````

## 5. Different objects, same class

````python
anna = Patient("Anna", 42, 37.0)
marco = Patient("Marco", 67, 38.4)

print(anna.has_fever())
print(marco.has_fever())
````

## 6. A useful representation: `__repr__`

````python
class Patient:
    def __init__(self, name, age, temperature):
        self.name = name
        self.age = age
        self.temperature = temperature

    def has_fever(self, threshold=38.0):
        return self.temperature >= threshold

    def __repr__(self):
        return (
            f"Patient(name={self.name!r}, "
            f"age={self.age}, "
            f"temperature={self.temperature})"
        )

patient = Patient("Anna", 42, 37.1)
print(patient)
````

## 7. Encapsulation

A class groups related data and behavior together.

For example, a `Rectangle` can store its dimensions and know how to compute its area.

````python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

rectangle = Rectangle(4, 3)

print(rectangle.area())
````

## 8. Properties

Properties let us expose computed values like attributes.

````python
class Circle:
    def __init__(self, radius):
        self.radius = radius

    @property
    def area(self):
        return 3.14159 * self.radius**2

circle = Circle(2)

print(circle.area)
````

## 9. Class attributes

Class attributes are shared by all instances.

````python
class Patient:
    fever_threshold = 38.0

    def __init__(self, name, temperature):
        self.name = name
        self.temperature = temperature

    def has_fever(self):
        return self.temperature >= Patient.fever_threshold

p = Patient("Anna", 38.2)

print(p.has_fever())
````

## 10. Inheritance

Inheritance allows one class to reuse and extend another class.

````python
class Person:
    def __init__(self, name):
        self.name = name

    def describe(self):
        return f"Person: {self.name}"


class Student(Person):
    def __init__(self, name, student_id):
        super().__init__(name)
        self.student_id = student_id

    def describe(self):
        return f"Student: {self.name}, ID={self.student_id}"


student = Student("Elena", "S001")

print(student.describe())
````

## 11. Composition

Often, composition is simpler than inheritance.

One object can contain another object.

````python
class Address:
    def __init__(self, city, country):
        self.city = city
        self.country = country


class Person:
    def __init__(self, name, address):
        self.name = name
        self.address = address


address = Address("Trieste", "Italy")
person = Person("Anna", address)

print(person.address.city)
````

## 12. When should you use classes?

Classes are useful when:

- you have many entities with the same structure;
- data and behavior naturally belong together;
- objects have state that changes over time;
- the abstraction makes the program easier to understand.

Classes are not automatically better than functions.

For simple transformations, functions may be clearer.

# Exercise

Create a class `MeasurementSeries` with:

- an attribute `values`;
- a method `mean()`;
- a method `minimum()`;
- a method `maximum()`;
- a property `count`.

Then create an object and test all methods.

````python
# Write your solution here
````

# Possible solution

````python
class MeasurementSeries:
    def __init__(self, values):
        self.values = values

    def mean(self):
        return sum(self.values) / len(self.values)

    def minimum(self):
        return min(self.values)

    def maximum(self):
        return max(self.values)

    @property
    def count(self):
        return len(self.values)


series = MeasurementSeries([10, 12, 15, 9])

print(series.mean())
print(series.minimum())
print(series.maximum())
print(series.count)
````

# Take-home messages

- A class defines a type of object.
- `__init__` initializes object attributes.
- `self` refers to the current object.
- Methods define behavior.
- Properties expose computed values like attributes.
- Inheritance can reuse behavior, but composition is often simpler.
- OOP is useful when data and behavior naturally belong together.
