# 5. Exploratory Data Analysis

Exploratory Data Analysis (EDA) is the process of understanding a dataset before building models or drawing conclusions.

EDA combines:

- numerical summaries;
- visualization;
- missing-value analysis;
- distribution analysis;
- relationships between variables;
- dimensionality reduction.

This notebook also introduces **Principal Component Analysis (PCA)** as an exploratory tool.

- interpret box plots and scatter plots;
- understand why scaling matters;
- use PCA for exploratory visualization.

## 1. Load the dataset

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("patient_data.csv")
df.head()
````

## 2. First inspection

Start with basic questions:

- How many rows and columns?
- What are the variable types?
- Are there missing values?
- Are the values plausible?

````python
print("Shape:", df.shape)
print()
df.info()
````

````python
df.describe(include="all")
````

## 3. Missing values

````python
missing = df.isna().sum().sort_values(ascending=False)
missing
````

Missing values are not automatically a problem, but they must be understood.

Possible questions:

- Why are values missing?
- Are they missing randomly?
- Is the missingness concentrated in particular groups?

## 4. Numerical distributions

````python
df["age"].hist(bins=20)

plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age distribution")
plt.show()
````

````python
df["bmi"].hist(bins=20)

plt.xlabel("BMI")
plt.ylabel("Frequency")
plt.title("BMI distribution")
plt.show()
````

## 5. Box plots

A box plot summarizes:

- median;
- first and third quartiles;
- interquartile range;
- potentially unusual observations.

````python
df.boxplot(column="cholesterol")

plt.ylabel("Cholesterol")
plt.title("Cholesterol distribution")
plt.show()
````

## 6. Categorical variables

````python
df["smoker"].value_counts()
````

````python
df["city"].value_counts().plot(kind="bar")

plt.ylabel("Count")
plt.title("Patients by city")
plt.show()
````

## 7. Relationship between two numerical variables

````python
plt.scatter(
    df["age"],
    df["systolic_bp"],
    alpha=0.6
)

plt.xlabel("Age")
plt.ylabel("Systolic blood pressure")
plt.title("Age vs systolic blood pressure")
plt.show()
````

## 8. Compare groups

````python
df.boxplot(
    column="systolic_bp",
    by="smoker"
)

plt.suptitle("")
plt.title("Blood pressure by smoking status")
plt.xlabel("Smoker")
plt.ylabel("Systolic blood pressure")
plt.show()
````

## 9. Correlation

Correlation measures linear association between numerical variables.

A value near:

- `+1` suggests strong positive linear association;
- `-1` suggests strong negative linear association;
- `0` suggests little linear association.

Important: zero correlation does **not** imply statistical independence.

````python
numeric_columns = [
    "age",
    "bmi",
    "systolic_bp",
    "cholesterol",
    "exercise_hours_week"
]

correlation_matrix = df[numeric_columns].corr()

correlation_matrix
````

````python
plt.imshow(
    correlation_matrix,
    aspect="auto"
)

plt.xticks(
    range(len(numeric_columns)),
    numeric_columns,
    rotation=45,
    ha="right"
)

plt.yticks(
    range(len(numeric_columns)),
    numeric_columns
)

plt.colorbar(label="Correlation")

plt.title("Correlation matrix")
plt.tight_layout()
plt.show()
````

# PCA as an EDA tool

## 10. What is PCA?

Principal Component Analysis creates new variables called **principal components**.

Each component is a linear combination of the original numerical variables.

The components are chosen so that:

1. the first component explains as much variance as possible;
2. the second explains as much as possible while being orthogonal to the first;
3. and so on.

PCA can help us:

- visualize high-dimensional data;
- detect dominant directions of variation;
- identify groups or unusual observations;
- understand redundancy between variables.

## 11. Why scaling matters

Suppose one variable is measured in years and another in hundreds of units.

A variable with a much larger numerical scale can dominate PCA.

Therefore, PCA is often applied after standardization:

```text
x_scaled = (x - mean) / standard deviation
```

````python
from sklearn.preprocessing import StandardScaler

pca_features = df[
    [
        "age",
        "bmi",
        "systolic_bp",
        "cholesterol",
        "exercise_hours_week"
    ]
].copy()

pca_features = pca_features.fillna(
    pca_features.median()
)

scaler = StandardScaler()

X_scaled = scaler.fit_transform(
    pca_features
)

X_scaled[:5]
````

## 12. Fit PCA

````python
from sklearn.decomposition import PCA

pca = PCA()

X_pca = pca.fit_transform(
    X_scaled
)

print(
    pca.explained_variance_ratio_
)
````

## 13. Explained variance

The explained variance ratio tells us how much of the total variability is captured by each component.

````python
explained = pca.explained_variance_ratio_

plt.plot(
    range(1, len(explained) + 1),
    explained,
    marker="o"
)

plt.xlabel("Principal component")
plt.ylabel("Explained variance ratio")
plt.title("PCA explained variance")
plt.show()
````

We can also look at cumulative explained variance.

````python
cumulative = np.cumsum(explained)

plt.plot(
    range(1, len(cumulative) + 1),
    cumulative,
    marker="o"
)

plt.xlabel("Number of components")
plt.ylabel("Cumulative explained variance")
plt.ylim(0, 1.05)
plt.title("Cumulative explained variance")
plt.show()
````

## 14. Two-dimensional PCA visualization

````python
pca_2d = PCA(n_components=2)

X_2d = pca_2d.fit_transform(
    X_scaled
)

pca_df = pd.DataFrame(
    X_2d,
    columns=["PC1", "PC2"]
)

pca_df["high_risk"] = df["high_risk"].values

for label in pca_df["high_risk"].unique():
    subset = pca_df[
        pca_df["high_risk"] == label
    ]

    plt.scatter(
        subset["PC1"],
        subset["PC2"],
        alpha=0.6,
        label=label
    )

plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PCA projection")
plt.legend(title="High risk")
plt.show()
````

The plot does not automatically prove that the groups are statistically distinct.

It is an exploratory visualization.

## 15. PCA loadings

The PCA components tell us how strongly each original variable contributes to a component.

````python
loadings = pd.DataFrame(
    pca_2d.components_.T,
    index=pca_features.columns,
    columns=["PC1", "PC2"]
)

loadings
````

Large absolute loading values indicate variables that contribute strongly to a principal component.

The sign indicates direction, but PCA signs can be flipped without changing the underlying solution.

# Take-home messages

- EDA should come before modeling.
- Use both numerical summaries and plots.
- Missing values and outliers require context, not automatic deletion.
- Correlation measures linear association, not general dependence.
- PCA can reveal dominant directions of variation.
- Scaling is usually important before PCA.
- PCA projections are exploratory and should not be over-interpreted.
