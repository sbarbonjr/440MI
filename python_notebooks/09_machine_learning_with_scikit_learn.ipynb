# 6. Python Programming with Machine Learning: The scikit-learn API

`scikit-learn` provides a consistent API for machine-learning workflows.

The core pattern is:

```python
model.fit(X_train, y_train)
predictions = model.predict(X_test)
```

## Learning objectives

By the end of this notebook, you should be able to:

- separate features and targets;
- split data into train and test sets;
- fit a scikit-learn estimator;
- make predictions;
- evaluate a classifier;
- use preprocessing;
- build a pipeline;
- understand the role of `fit`, `transform`, and `predict`.

## 1. The estimator API

Most scikit-learn objects follow a common interface.

### Estimator

```python
model.fit(X, y)
```

### Predictor

```python
model.predict(X)
```

### Transformer

```python
transformer.fit(X)
X_new = transformer.transform(X)
```

Many transformers also support:

```python
X_new = transformer.fit_transform(X)
```

## 2. Load a built-in dataset

````python
from sklearn.datasets import load_breast_cancer

data = load_breast_cancer(
    as_frame=True
)

X = data.data
y = data.target

X.head()
````

````python
y.value_counts()
````

## 3. Train/test split

````python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

print(X_train.shape)
print(X_test.shape)
````

The test set should remain unseen during training.

## 4. Fit a simple model

````python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(
    max_iter=5000
)

model.fit(
    X_train,
    y_train
)
````

## 5. Make predictions

````python
y_pred = model.predict(
    X_test
)

y_pred[:10]
````

## 6. Evaluate classification accuracy

````python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(
    y_test,
    y_pred
)

print("Accuracy:", accuracy)
````

Accuracy is not always sufficient, especially for imbalanced datasets.

## 7. Confusion matrix

````python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(
    y_test,
    y_pred
)

cm
````

## 8. Classification report

````python
from sklearn.metrics import classification_report

print(
    classification_report(
        y_test,
        y_pred
    )
)
````

## 9. Why preprocessing matters

Many models benefit from features on comparable scales.

A common approach is standardization.

````python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(
    X_train
)

X_test_scaled = scaler.transform(
    X_test
)
````

Important:

- `fit()` the scaler on the training set;
- use the fitted scaler to `transform()` the test set.

Do not fit preprocessing separately on the test set.

## 10. Fit the model on scaled data

````python
scaled_model = LogisticRegression(
    max_iter=5000
)

scaled_model.fit(
    X_train_scaled,
    y_train
)

scaled_predictions = scaled_model.predict(
    X_test_scaled
)

print(
    accuracy_score(
        y_test,
        scaled_predictions
    )
)
````

## 11. Pipelines

A pipeline chains preprocessing and modeling steps.

This reduces mistakes and keeps the workflow compact.

````python
from sklearn.pipeline import Pipeline

pipeline = Pipeline(
    steps=[
        ("scaler", StandardScaler()),
        ("model", LogisticRegression(max_iter=5000))
    ]
)

pipeline.fit(
    X_train,
    y_train
)

pipeline_predictions = pipeline.predict(
    X_test
)

print(
    accuracy_score(
        y_test,
        pipeline_predictions
    )
)
````

## 12. Another estimator: k-nearest neighbors

````python
from sklearn.neighbors import KNeighborsClassifier

knn_pipeline = Pipeline(
    steps=[
        ("scaler", StandardScaler()),
        ("model", KNeighborsClassifier(n_neighbors=5))
    ]
)

knn_pipeline.fit(
    X_train,
    y_train
)

knn_predictions = knn_pipeline.predict(
    X_test
)

print(
    accuracy_score(
        y_test,
        knn_predictions
    )
)
````

## 13. Decision trees

````python
from sklearn.tree import DecisionTreeClassifier

tree = DecisionTreeClassifier(
    max_depth=4,
    random_state=42
)

tree.fit(
    X_train,
    y_train
)

tree_predictions = tree.predict(
    X_test
)

print(
    accuracy_score(
        y_test,
        tree_predictions
    )
)
````

## 14. Model parameters vs learned attributes

Before fitting:

````python
print(
    tree.get_params()
)
````

After fitting, learned attributes often end with an underscore.

Examples include:

```python
classes_
feature_importances_
coef_
```

````python
tree.feature_importances_[:10]
````

## 15. Predict probabilities

````python
probabilities = pipeline.predict_proba(
    X_test
)

probabilities[:5]
````

Probabilities can be useful when the decision threshold matters.

# Take-home messages

- scikit-learn uses a consistent estimator API.
- `fit()` learns from training data.
- `transform()` changes feature representation.
- `predict()` produces outputs.
- Keep test data separate from training.
- Fit preprocessing only on the training data.
- Pipelines reduce preprocessing mistakes.
- Different estimators can often be swapped with minimal code changes.
