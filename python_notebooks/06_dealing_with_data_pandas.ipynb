## 4. Load a CSV file

This notebook uses the file `patient_data.csv` included with the course materials.

````python
patients = pd.read_csv("patient_data.csv")
patients.head()
````

## 5. Select columns

````python
patients["age"].head()
````

````python
patients[["age", "bmi", "systolic_bp"]].head()
````

## 6. Select rows with `loc` and `iloc`

````python
patients.loc[0:4, ["age", "bmi"]]
````

````python
patients.iloc[0:5, 0:4]
````

## 7. Filtering rows

````python
older_patients = patients[
    patients["age"] >= 65
]

older_patients.head()
````

Multiple conditions require parentheses.

````python
subset = patients[
    (patients["age"] >= 65)
    & (patients["smoker"] == "yes")
]

subset.head()
````

## 8. Create new columns

````python
patients["pulse_pressure_proxy"] = (
    patients["systolic_bp"] - 80
)

patients.head()
````

## 9. Missing values

````python
patients.isna().sum()
````

Drop rows with missing values:

````python
complete_cases = patients.dropna()
print(complete_cases.shape)
````

Fill missing values:

````python
patients["bmi_filled"] = patients["bmi"].fillna(
    patients["bmi"].median()
)

patients[["bmi", "bmi_filled"]].head(10)
````

## 10. Sorting

````python
patients.sort_values(
    "cholesterol",
    ascending=False
).head()
````

## 11. Grouping and aggregation

````python
patients.groupby("smoker")["systolic_bp"].mean()
````

````python
patients.groupby("city").agg(
    mean_age=("age", "mean"),
    mean_bmi=("bmi", "mean"),
    number_of_patients=("patient_id", "count")
)
````

## 12. Value counts

````python
patients["smoker"].value_counts()
````

````python
patients["city"].value_counts(normalize=True)
````

## 13. Rename columns

````python
renamed = patients.rename(
    columns={
        "systolic_bp": "systolic_blood_pressure"
    }
)

renamed.head()
````

## 14. Apply a function

Use vectorized pandas operations when possible.

For simple transformations, `apply()` is available.

````python
def age_group(age):
    if age < 40:
        return "young"
    elif age < 65:
        return "middle"
    else:
        return "older"

patients["age_group"] = patients["age"].apply(age_group)

patients[["age", "age_group"]].head()
````

## 15. Export data

````python
patients.to_csv(
    "patient_data_processed.csv",
    index=False
)
````

# Possible solution

````python
df = pd.read_csv("patient_data.csv")

filtered = df[df["age"] > 50].copy()

filtered["high_bp"] = (
    filtered["systolic_bp"] >= 140
)

print(
    filtered.groupby("smoker")["bmi"].mean()
)

filtered.to_csv(
    "patients_over_50.csv",
    index=False
)
````
