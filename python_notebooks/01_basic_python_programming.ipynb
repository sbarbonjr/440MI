## 1. Your first Python program

````python
print("First instruction")
print("Second instruction")
print("Third instruction")
````

## 2. Comments

Comments start with the `#` symbol.

Python ignores everything after `#` on the same line.

Comments are useful to explain **why** the code does something.

````python
# This is a comment

temperature = 37.2

print(temperature)
````

### Good practice

Avoid comments that simply repeat what the code already says.

Not very useful:

```python
# Add 1 to x
x = x + 1
```

Comments are more useful when they explain a non-obvious choice or assumption.

## 3. Variables

A variable associates a **name** with a value.

```python
variable_name = value
```

The symbol `=` means **assignment**.

It does not mean mathematical equality.

````python
age = 45
temperature = 36.8
patient_name = "Anna"

print(age)
print(temperature)
print(patient_name)
````

A variable can change value during program execution.

````python
score = 10
print(score)

score = 15
print(score)
````

This is perfectly valid:

```python
x = x + 1
```

It means:

1. read the current value of `x`;
2. add `1`;
3. store the new value back in `x`.

````python
x = 5
x = x + 1

print(x)
````

## 4. Variable names

A variable name:

- can contain letters, digits, and `_`;
- cannot start with a digit;
- cannot contain spaces;
- is **case-sensitive**.

Therefore, `age` and `Age` are two different variables.

### Good practice: use `snake_case`

Python commonly uses names such as:

```python
patient_age
blood_pressure
number_of_samples
```

rather than:

```python
PatientAge
bloodpressure
n
```

when the meaning is not obvious.

````python
patient_age = 62
body_temperature = 37.1
number_of_visits = 4

print(patient_age)
````

### Prefer descriptive names

Better:

```python
mean_temperature = 36.9
```

than:

```python
mt = 36.9
```

Short names such as `i`, `j`, or `x` are still acceptable when their meaning is obvious from context.

## 5. Basic data types

Some of the most common Python data types are:

| Type | Example | Meaning |
|---|---|---|
| `int` | `42` | integer number |
| `float` | `3.14` | decimal number |
| `str` | `"hello"` | text |
| `bool` | `True` | logical true/false value |

The function `type()` tells us the type of a value.

````python
age = 45
temperature = 36.7
name = "Marco"
is_smoker = False

print(type(age))
print(type(temperature))
print(type(name))
print(type(is_smoker))
````

## 6. Numbers and arithmetic operators

Python can be used as a calculator.

````python
a = 10
b = 3

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Integer division:", a // b)
print("Remainder:", a % b)
print("Power:", a ** b)
````

### Operator precedence

As in mathematics:

1. parentheses;
2. powers;
3. multiplication and division;
4. addition and subtraction.

Parentheses are often useful even when they are not strictly required, because they make code easier to read.

````python
result_1 = 2 + 3 * 4
result_2 = (2 + 3) * 4

print(result_1)
print(result_2)
````

## 7. Strings

Strings represent text.

You can use either single or double quotes.

````python
first_name = "Anna"
last_name = 'Rossi'

print(first_name)
print(last_name)
````

Strings can be concatenated with `+`.

````python
full_name = first_name + " " + last_name

print(full_name)
````

### f-strings

A convenient way to insert variables into text is to use an **f-string**.

````python
age = 34
name = "Anna"

print(f"{name} is {age} years old.")
````

## 8. Boolean values and comparisons

A Boolean value can be:

```python
True
False
```

Comparison operators produce Boolean values.

````python
age = 70

print(age > 65)
print(age < 65)
print(age == 70)
print(age != 70)
print(age >= 18)
````

### Important: `=` versus `==`

```python
x = 5
```

assigns the value `5` to `x`.

```python
x == 5
```

checks whether `x` is equal to `5`.

## 9. Logical operators

We can combine conditions using:

- `and`
- `or`
- `not`

````python
age = 72
has_symptoms = True

high_risk = age > 65 and has_symptoms

print(high_risk)
````

````python
temperature = 38.2

has_fever = temperature >= 38
print(has_fever)

print(not has_fever)
````

## 13. A first look at lists

A list stores multiple values in a single object.

````python
temperatures = [36.5, 37.0, 38.1, 36.9]

print(temperatures)
````

List indices start from `0`.

````python
print(temperatures[0])
print(temperatures[1])
print(temperatures[-1])
````

Lists are mutable, so they can be changed.

````python
temperatures.append(37.4)

print(temperatures)
````

Useful built-in functions include:

````python
print("Number of elements:", len(temperatures))
print("Minimum:", min(temperatures))
print("Maximum:", max(temperatures))
print("Sum:", sum(temperatures))
print("Mean:", sum(temperatures) / len(temperatures))
````

## 14. A first look at dictionaries

A dictionary stores **key-value pairs**.

````python
patient = {
    "name": "Anna",
    "age": 54,
    "temperature": 37.8
}

print(patient)
print(patient["name"])
print(patient["age"])
````

Dictionary values can be updated or extended.

````python
patient["temperature"] = 38.0
patient["smoker"] = False

print(patient)
````

## 15. Functions

Functions group instructions into reusable units.

They are defined using `def`.

````python
def greet():
    print("Hello!")

greet()
````

Functions can accept parameters.

````python
def greet(name):
    print(f"Hello, {name}!")

greet("Anna")
greet("Marco")
````

Functions can return values using `return`.

````python
def calculate_bmi(weight_kg, height_m):
    bmi = weight_kg / height_m**2
    return bmi

result = calculate_bmi(70, 1.75)

print(result)
````

### Good function design

A function should ideally:

- do one clearly defined task;
- have a descriptive name;
- avoid unnecessary complexity;
- reduce repeated code.

## 17. Common errors

Errors are a normal part of programming.

Learning to read an error message is an essential skill.

### `NameError`

This occurs when a variable does not exist.

Example:

```python
print(patient_age)
```

if `patient_age` was never defined.

### `TypeError`

This occurs when an operation is applied to incompatible types.

Example:

```python
"Age: " + 42
```

Possible solutions:

```python
"Age: " + str(42)
```

or:

```python
f"Age: {42}"
```

### `IndexError`

This occurs when a list index does not exist.

```python
values = [10, 20, 30]
values[5]
```

## 18. Good programming practices

### 1. Use clear names

Prefer:

```python
mean_temperature = 37.1
```

over:

```python
x = 37.1
```

when the meaning of `x` is not obvious.

### 2. Use `snake_case`

Examples:

```python
patient_age
blood_pressure
mean_value
```

### 3. Add spaces around operators

Prefer:

```python
mean = total / number_of_samples
```

over:

```python
mean=total/number_of_samples
```

### 4. Avoid magic numbers

If a value has a clear meaning and is reused, give it a name.

````python
FEVER_THRESHOLD = 38.0

temperature = 38.2

if temperature >= FEVER_THRESHOLD:
    print("Fever")
````

### 5. Do not repeat yourself

A common principle is:

**DRY = Don't Repeat Yourself**

If the same logic appears several times, it may belong in a function.

````python
def calculate_bmi(weight_kg, height_m):
    return weight_kg / height_m**2

print(calculate_bmi(70, 1.75))
print(calculate_bmi(62, 1.68))
````

### 6. Keep code simple

Readable code is usually preferable to unnecessarily compact code.

````python
age = 72
has_diabetes = True

is_high_risk = age > 65 and has_diabetes

print(is_high_risk)
````

## 19. Mini example

Suppose we have temperature values from several patients.

We want to:

1. compute the mean temperature;
2. count how many patients have temperature at least `38 °C`;
3. print a summary.

````python
temperatures = [36.7, 38.2, 37.1, 39.0, 36.5]

FEVER_THRESHOLD = 38.0

mean_temperature = sum(temperatures) / len(temperatures)

number_with_fever = 0

for temperature in temperatures:
    if temperature >= FEVER_THRESHOLD:
        number_with_fever += 1

print(f"Mean temperature: {mean_temperature:.1f} °C")
print(f"Patients with fever: {number_with_fever}")
````

The line:

```python
number_with_fever += 1
```

is a shorter form of:

```python
number_with_fever = number_with_fever + 1
```
